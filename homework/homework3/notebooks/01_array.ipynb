NumPy Operations

```python
import numpy as np
import time

arr = np.array([1, 2, 3, 4, 5])
print("Array:", arr)
print("Plus 10:", arr + 10)


large_array = np.arange(1_000_000)
# Loop version
start = time.time()
loop_result = [x * 2 for x in large_array]
loop_time = time.time() - start
# Vectorized version
start = time.time()
vec_result = large_array * 2
vec_time = time.time() - start

print(f"Loop time: {loop_time:.5f} s")
print(f"Vectorized time: {vec_time:.5f} s")
```
